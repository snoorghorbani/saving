# Split Savings to Meet Weekly Goal

This notebook helps you calculate how to distribute your current total savings over weeks to ensure each week meets a specified goal (default: 385 OMR).

## 1. Input Total Savings and Goal

Enter your current total savings and the weekly goal amount (default: 385 OMR).

```python
# Input your current total savings and weekly goal
try:
    total_savings = float(input('Enter your current total savings (OMR): '))
except:
    total_savings = 0.0
    print('Invalid input, defaulting to 0.0')
try:
    weekly_goal = float(input('Enter your weekly goal (OMR, default 385): ') or 385)
except:
    weekly_goal = 385.0
    print('Invalid input, defaulting to 385.0')
print(f"Total savings: {total_savings} OMR, Weekly goal: {weekly_goal} OMR")
```

## 2. Calculate Weekly Savings Needed

Given your total savings and weekly goal, calculate how many weeks you can cover and how to distribute the savings.

```python
import math

# Calculate how many full weeks can be covered by the savings
total_weeks = int(total_savings // weekly_goal)
remainder = total_savings - (total_weeks * weekly_goal)

print(f"You can fully cover {total_weeks} weeks with {weekly_goal} OMR each.")
if remainder > 0:
    print(f"Remaining savings: {remainder:.3f} OMR (can be distributed or added to the last week)")
```

## 3. Determine Weeks to Reach Goal

If you want to reach a target total by saving a fixed amount per week, calculate how many weeks it will take.

```python
# Example: If you want to reach a target total by saving weekly
try:
    target_total = float(input('Enter your target total savings (OMR): ') or (weekly_goal * 52))
except:
    target_total = weekly_goal * 52
    print('Invalid input, defaulting to 1 year of weekly goal')
try:
    weekly_saving = float(input('Enter your planned weekly saving (OMR): ') or weekly_goal)
except:
    weekly_saving = weekly_goal
    print('Invalid input, defaulting to weekly goal')

weeks_needed = math.ceil((target_total - total_savings) / weekly_saving) if total_savings < target_total else 0
print(f"Weeks needed to reach {target_total} OMR: {weeks_needed}")
```

## 4. Visualize Savings Progress

Plot a graph showing cumulative savings over weeks compared to the goal.

```python
import matplotlib.pyplot as plt
import numpy as np

# Simulate weekly savings distribution
distribution = [weekly_goal] * total_weeks
if remainder > 0:
    distribution.append(remainder)

cumulative = np.cumsum(distribution)
weeks = np.arange(1, len(cumulative) + 1)
goal_line = [weekly_goal * w for w in weeks]

plt.figure(figsize=(10, 5))
plt.plot(weeks, cumulative, label='Cumulative Savings', marker='o')
plt.plot(weeks, goal_line, label='Goal Line', linestyle='--')
plt.xlabel('Week')
plt.ylabel('OMR')
plt.title('Cumulative Savings vs. Goal')
plt.legend()
plt.grid(True)
plt.show()
```
